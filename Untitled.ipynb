## Load GEO Series Matrix file

```python
import pandas as pd

# Load only the expression matrix part
ML_file_path = r"C:\Users\gargi\Desktop\Publication\Transcriptomic_and_Epigenomic_Features\GSE9006-GPL96_series_matrix.txt"

# Read the file line-by-line first to find where the data starts
with open(ML_file_path, 'r') as f:
    lines = f.readlines()

# Find where the expression data starts
start_idx = [i for i, line in enumerate(lines) if "!series_matrix_table_begin" in line][0] + 1
end_idx = [i for i, line in enumerate(lines) if "!series_matrix_table_end" in line][0]

# Read only the data portion into a DataFrame
from io import StringIO
data_str = ''.join(lines[start_idx:end_idx])
df = pd.read_csv(StringIO(data_str), sep='\t', index_col=0)

# Show preview
df.head()
```

## Get the labels (T2D vs control)

```python
# Extract sample annotations
sample_lines = [line for line in lines if line.startswith("!Sample_characteristics_ch1")]

# Check what metadata is available
for i, line in enumerate(sample_lines[:5]):
    print(f"Sample {i+1}: {line.strip()}")
```

## Extract sample labels from metadata

```python
# Extract sample labels
sample_lines = [line for line in lines if line.startswith("!Sample_characteristics_ch1")]

# Clean and organize labels
labels = [line.strip().split('\t')[1:] for line in sample_lines]

# Transpose the label matrix so each list is per sample
import pandas as pd
labels_df = pd.DataFrame(labels).T
labels_df.columns = [f"characteristic_{i}" for i in range(labels_df.shape[1])]
labels_df.index = df.columns  # Match the sample names

labels_df.head()
```

## Identify the column containing disease state

```python
for col in labels_df.columns:
    print(f"\n{col}:\n", labels_df[col].value_counts())
```

## Create binary labels (0 = control, 1 = T2D)

```python
# Extract illness column
illness_col = labels_df['characteristic_4']

# Clean and encode illness
y = illness_col.apply(
    lambda x: 1 if "diabetes" in x.lower() else 0
)

# Preview
y.value_counts()
```

## Split Data

```python
from sklearn.model_selection import train_test_split

X = df.T
y = labels_df['characteristic_4'].apply(lambda x: 1 if "diabetes" in x.lower() else 0)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, stratify=y, test_size=0.2, random_state=42
)
```

## Handle Class Imbalance

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

clf = RandomForestClassifier(class_weight='balanced', random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

print(classification_report(y_test, y_pred))

```

## Issues:
### Class imbalance: Only 5 control samples vs. 19 T2D — the model is biased toward predicting T2D.

### Precision/Recall for class 0 = 0: Your Random Forest never correctly predicted any control sample.

### Macro average is poor (0.40–0.50): Indicates poor balance between classes.
