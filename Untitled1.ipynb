## Load GEO Series Matrix file

```python
import pandas as pd

# Load only the expression matrix part
ML_file_path = r"C:\Users\gargi\Desktop\Publication\Transcriptomic_and_Epigenomic_Features\GSE9006-GPL96_series_matrix.txt"

# Read the file line-by-line first to find where the data starts
with open(ML_file_path, 'r') as f:
    lines = f.readlines()

# Find where the expression data starts
start_idx = [i for i, line in enumerate(lines) if "!series_matrix_table_begin" in line][0] + 1
end_idx = [i for i, line in enumerate(lines) if "!series_matrix_table_end" in line][0]

# Read only the data portion into a DataFrame
from io import StringIO
data_str = ''.join(lines[start_idx:end_idx])
df = pd.read_csv(StringIO(data_str), sep='\t', index_col=0)

# Show preview
df.head()
```

## Get the labels (T2D vs control)

```python
# Extract sample annotations
sample_lines = [line for line in lines if line.startswith("!Sample_characteristics_ch1")]

# Check what metadata is available
for i, line in enumerate(sample_lines[:5]):
    print(f"Sample {i+1}: {line.strip()}")
```

## Extract sample labels from metadata

```python
# Extract sample labels
sample_lines = [line for line in lines if line.startswith("!Sample_characteristics_ch1")]

# Clean and organize labels
labels = [line.strip().split('\t')[1:] for line in sample_lines]

# Transpose the label matrix so each list is per sample
import pandas as pd
labels_df = pd.DataFrame(labels).T
labels_df.columns = [f"characteristic_{i}" for i in range(labels_df.shape[1])]
labels_df.index = df.columns  # Match the sample names

labels_df.head()
```

## Identify the column containing disease state

```python
for col in labels_df.columns:
    print(f"\n{col}:\n", labels_df[col].value_counts())
```

## Create binary labels (0 = control, 1 = T2D)

```python
# Extract illness column
illness_col = labels_df['characteristic_4']

# Clean and encode illness
y = illness_col.apply(
    lambda x: 1 if "diabetes" in x.lower() else 0
)

# Preview
y.value_counts()
```

## Split Data

```python
from sklearn.model_selection import train_test_split

X = df.T
y = labels_df['characteristic_4'].apply(lambda x: 1 if "diabetes" in x.lower() else 0)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, stratify=y, test_size=0.2, random_state=42
)
```

## Handle Class Imbalance

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

clf = RandomForestClassifier(class_weight='balanced', random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

print(classification_report(y_test, y_pred, zero_division=0))
```

## Issues:
### Class imbalance: Only 5 control samples vs. 19 T2D — the model is biased toward predicting T2D.

### Precision/Recall for class 0 = 0: Your Random Forest never correctly predicted any control sample.

### Macro average is poor (0.40–0.50): Indicates poor balance between classes.


```python
# Total number of samples and features
print(f"Total samples: {X.shape[0]}")
print(f"Total features: {X.shape[1]}")

# Distribution of target classes
print("\nClass distribution:")
print(y.value_counts())
```

## The classes are imbalanced, with the minority class (control) making up ~20% of the data.

## Apply SMOTE

```python
# Check shapes
print("Original shape:", X.shape)
print("Class distribution:\n", y.value_counts())
```

SMOTE (Synthetic Minority Over-sampling Technique) creates synthetic examples of the minority class by interpolating between existing samples. However:

High dimensionality (22,283 features) and

Small number of minority samples (only 24)

can cause problems like:

Sparse feature space → Interpolated samples may not be meaningful.

n_neighbors issue → SMOTE requires at least k+1 samples in the minority class, where k is usually 5 by default. With only 24 samples, this is tight.

Risk of overfitting due to synthetic noise in such high-dimensional data.



```python
from imblearn.over_sampling import SMOTE

# Use k_neighbors=3 due to low number of minority class samples
smote = SMOTE(k_neighbors=3, random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)
```

```python
import pandas as pd
print("Resampled class distribution:\n", pd.Series(y_resampled).value_counts())
```

#Train-test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled, test_size=0.2, random_state=42, stratify=y_resampled)
```

## Model training

```python
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier(random_state=42)
clf.fit(X_train, y_train)
```

## Model evaluation

```python
from sklearn.metrics import classification_report, confusion_matrix
y_pred = clf.predict(X_test)
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

Balanced classes after SMOTE + strong generalization on unseen data.

Very low false positives/negatives — only one class 0 sample was misclassified.

High precision + recall for both classes → great for applications like disease risk prediction or rare event classification.

## Cross-validation for robustness

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(clf, X_resampled, y_resampled, cv=5)
print("CV Accuracy Scores:", scores)
print("Mean Accuracy:", scores.mean())
```

```python
feature_importances = pd.Series(clf.feature_importances_, index=feature_names)
print(feature_importances.sort_values(ascending=False).head(10))
```
